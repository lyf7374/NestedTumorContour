```python
import os
import torch
import h5py
import numpy as np
import tqdm
import torch
# Original directory containing .pt files
pt_data_dir = '/Users/yifanli/Desktop/dataset/AfterCT_BraTS20/preprocessed_data_BraTS20'

# New directory to save compressed .h5 files
h5_data_dir = 'preprocessed_data_BraTS20_h5'

if not os.path.exists(h5_data_dir):
    os.makedirs(h5_data_dir)

# List all .pt files in the original directory
pt_files = sorted([f for f in os.listdir(pt_data_dir) if f.endswith('.pt')])

# Function to optimize data types and save as HDF5
def save_to_hdf5(group, data):
    if isinstance(data, dict):
        for key, value in data.items():
            if isinstance(value, np.ndarray):
                # Save numpy arrays directly as datasets
                array = value
                if array.dtype == np.float64:
                    array = array.astype(np.float32)
                elif array.dtype == np.int64:
                    array = array.astype(np.int32)
                group.create_dataset(key, data=array, compression='gzip', compression_opts=9)
            elif isinstance(value, (np.number, float, int, str, bytes)):
                # Save basic data types as attributes
                group.attrs[key] = value
            else:
                # Recursively save nested structures
                subgroup = group.create_group(str(key))
                save_to_hdf5(subgroup, value)
    elif isinstance(data, (list, tuple)):
        # Handle lists and tuples
        for idx, item in enumerate(data):
            item_name = f'item_{idx}'
            if isinstance(item, np.ndarray):
                array = item
                if array.dtype == np.float64:
                    array = array.astype(np.float32)
                elif array.dtype == np.int64:
                    array = array.astype(np.int32)
                group.create_dataset(
                    item_name, data=array, compression='gzip', compression_opts=9
                )
            elif isinstance(item, (np.number, float, int, str, bytes)):
                group.attrs[item_name] = item
            else:
                subgroup = group.create_group(item_name)
                save_to_hdf5(subgroup, item)
    elif isinstance(data, np.ndarray):
        # Save numpy arrays
        array = data
        if array.dtype == np.float64:
            array = array.astype(np.float32)
        elif array.dtype == np.int64:
            array = array.astype(np.int32)
        group.create_dataset('data', data=array, compression='gzip', compression_opts=9)
    elif isinstance(data, (np.number, float, int, str, bytes)):
        # Save basic data types as attributes
        group.attrs['value'] = data
    else:
        # Convert unsupported types to string
        group.attrs['value'] = str(data)

def convert_pt_to_h5(pt_file_path, h5_file_path):
    # Load the data from the .pt file
    data = torch.load(pt_file_path)

    with h5py.File(h5_file_path, 'w') as hf:
        save_to_hdf5(hf, data)


```

```python
# 3. Define a desired batch size
batch_size = 2

# 4. Containers to hold data while batching
batch_imgs = []     # will hold NumPy arrays of shape [4, D, H, W]
batch_files = []    # will hold paths/identifiers

# 5. Loop through patient files
for pt_file in tqdm.tqdm(pt_files):
    pt_file_path = os.path.join(pt_data_dir, pt_file)

    # Load the dictionary
    data = torch.load(pt_file_path)
    
    # Extract and stack the modalities (shape [4, D, H, W])
    patient_image = np.stack([
        data['img']['t1'],
        data['img']['t1gd'],
        data['img']['t2'],
        data['img']['flair_data']
    ], axis=0)

    # Append to our batch lists
    batch_imgs.append(patient_image)
    batch_files.append(pt_file_path)

    # If we've reached the desired batch size, process them
    if len(batch_imgs) == batch_size:
        # Convert the list of [4, D, H, W] images to a float32 Tensor [B, 4, D, H, W]
        batch_tensor = torch.from_numpy(np.array(batch_imgs, dtype=np.float32))
        
        # Optional: move to GPU if available
        # batch_tensor = batch_tensor.cuda()
        # model = model.cuda()

        with torch.no_grad():
            # Get embeddings from model
            embeddings = model(batch_tensor)
            # embeddings shape depends on your forward pass. 
            # For instance, it might be [B, embed_dim] or something similar.

        # Save embeddings back into each file
        for i, path in enumerate(batch_files):
            file_data = torch.load(path)
            # Convert embedding to NumPy if needed
            file_data['img_embedding'] = embeddings[i].cpu().numpy()
            torch.save(file_data, path)

        # Reset batch storage
        batch_imgs = []
        batch_files = []

# 6. Handle any leftover files if the total number is not divisible by batch_size
if len(batch_imgs) > 0:
    batch_tensor = torch.from_numpy(np.array(batch_imgs, dtype=np.float32))
    # batch_tensor = batch_tensor.cuda()   # if you are using GPU
    with torch.no_grad():
        embeddings = model(batch_tensor)
    for i, path in enumerate(batch_files):
        file_data = torch.load(path)
        file_data['img_embedding'] = embeddings[i].cpu().numpy()
        torch.save(file_data, path)
```

```python
h5_data_dir = "/Users/yifanli/Desktop/fromLinux/TumorContour/preprocessed_data_BraTS20_emb_h5"

# New folder to save the "lightweight" .h5 files
h5_output_path = "light_h5_data"
if not os.path.exists(h5_output_path):
    os.makedirs(h5_output_path)

# Initialize the dataset
dataset = HDF5BrainDataset(h5_data_dir)

for i in tqdm.tqdm(range(len(dataset))):
    data_dict = dataset[i]  # entire dictionary from .h5

    # We'll build a "reduced" dictionary that keeps:
    #   data_dict['cts'][any_subkey]['item_1']
    #   data_dict['cts'][any_subkey]['item_2']
    #   data_dict['img_embedding'] (top-level)
    reduced_dict = {}
    cts_section = data_dict.get('cts', {})

    # cts_section is like:
    # {
    #   'TC_l': {'item_0':..., 'item_1':..., 'item_2':...},
    #   'TC_s': {...}, etc.
    # }
    new_cts = {}
    for subkey, subdict in cts_section.items():
        # subdict might have 'item_0', 'item_1', 'item_2', ...
        filtered_subdict = {}
        if 'item_1' in subdict:
            filtered_subdict['item_1'] = subdict['item_1']
        if 'item_2' in subdict:
            filtered_subdict['item_2'] = subdict['item_2']
        # Only add subkey if it has item_1 or item_2
        if filtered_subdict:
            new_cts[subkey] = filtered_subdict

    if new_cts:
        reduced_dict['cts'] = new_cts

    # Now, copy 'img_embedding' if it exists at top-level
    if 'img_embedding' in data_dict:
        reduced_dict['img_embedding'] = data_dict['img_embedding']

    # Name for the new .h5 file
    original_filename = dataset.data_files[i]  # e.g., "file_10.h5"
    new_file_path = os.path.join(h5_output_path, original_filename)

    # Write only the reduced data
    with h5py.File(new_file_path, 'w') as hf:
        save_to_hdf5(hf, reduced_dict)

print("All done! Reduced .h5 files are in:", h5_output_path)
```
